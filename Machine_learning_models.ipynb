#a. **Loss**
Loss is the difference between the predicted labels and the actual labels.

It describes how wrong model's predictions are.

**Diatance of loss**

In ML and statistics, loss measures the difference between predicted and actual values.

    For example if a model predicts 2 but the actual value is 5, we dont care that the loss is -3.
        2-5=-3
    when calculating loss, the (-) sign is assumed.




**Types of loss**

| Loss type             | Definition                                  | Formula                                     |
| :-------------------- | :------------------------------------------ | :------------------------------------------ |
| l1 loss               | The sum of the absolute values of the loss  | ∑|actual-predicted value|                    |
| Mean Absolute Error (MAE) | The average of L1                           | 1/N∑|actual-predicted value|              |
| L2 Loss               | The sum of the squared loss                 | ∑(actual-predicted value)²                  |
| Mean Squared Error (MSE) | The average of L2                           | 1/N∑(actual-predicted value)²             |
| Root Mean Squared Error (RMSE) | The square root of the mean squared error | $$\sqrt{1/N∑(actual-predicted value)²}$$ |


**LInear regression Equation**

we will use a common equation used in linear equation.

    y = mx+c

    where

    y: The dependent variable. It represents the vertical position on the graph.
    
    x: The independent variable. It represents the horizontal position.
    
    m: The slope (or gradient). This tells you how steep the line is.
    
        If m is positive, the line goes up.
        If m is negative, the line goes down.
        c: The y-intercept. This is the point where the line crosses the vertical y-axis (where x = 0).

In ML we writhe the equation for a linear regresion model as follows:

    y = b + W₁X₁

    where:

      y: is the predicted label

      b(w₀): is the bias. the same concept as the y-intercept in the algebric equation for a line.
      In ML sometimes refered to as w₀

      w₁: is the weight of the feature, the same concept as the slope m in the algebric equation for a line.

      x₁ is a feature, the input
    During training, the model calculates the weight and bias that produce the best model.

# Calculating loss example

Let's use the table below for our examples:

| x | actual (y) | predicted |
|---|------------|-----------|
| 1 | 4          | 4         |
| 2 | 3          | 2         |
| 3 | 2          | 1         |


### 1. L1 Loss

L1 Loss is the sum of the absolute differences between actual and predicted values.

$$\sum|actual - predicted|$$

For the given table:

*   $|4 - 4| = 0$
*   $|3 - 2| = 1$
*   $|2 - 1| = 1$

Total L1 Loss: $0 + 1 + 1 = 2$


### 2. L2 Loss
This is the sum of the squared of the error.

$$\sum(actual - predicted)^2$$

For the given table:

*   $(4 - 4)^2 = 0$
*   $(3 - 2)^2 = 1$
*   $(2 - 1)^2 = 1$

Total L2 Loss: $0 + 1 + 1 = 2$


### 3. Mean Absolute Error (MAE)
This is the average of L1 Loss.

$$\frac{1}{N}\sum|actual - predicted|$$

For the given table:

*   $|4 - 4| = 0$
*   $|3 - 2| = 1$
*   $|2 - 1| = 1$

Total MAE: $(0 + 1 + 1) / 3 = 2/3$


### 4. Mean Squared Error (MSE)
This is the average of L2 Loss across a set of N examples.

$$\frac{1}{N}\sum(actual - predicted)^2$$

For the given table:

*   $(4 - 4)^2 = 0$
*   $(3 - 2)^2 = 1$
*   $(2 - 1)^2 = 1$

Total MSE: $(0 + 1 + 1) / 3 = 2/3$


### 5. Root Mean Squared Error (RMSE)
This is the square root of the mean squared error.

$$\sqrt{\frac{1}{N}\sum(actual - predicted)^2}$$

For the given table:

*   $(4 - 4)^2 = 0$
*   $(3 - 2)^2 = 1$
*   $(2 - 1)^2 = 1$

Total RMSE: $\sqrt{(0 + 1 + 1) / 3} = \sqrt{2/3}$

# NB: MAE & RMSE can differ quite widely.

    MAE represents the the avg prediction error while RMSE reps the spread of the errors and is more skewed by larger errors

# **CHOOSING LOSS**

**Outliers** refer to how far a model's predictions are from the real value. In simple terms, it is a data point, that differs significantly from the rest of a group, falling far outside the normal pattern, average, or expected range. It represents an extreme value, anomaly, or exception that lies detached from the main cluster or body.

So decideing on whether to choose between **MAE** or **MSE** depends on the dataset you want to handle the prediction.

When choosing the best loss function, consider how you want the model to treat outliers.

For instance, MSE moves the model more toward the outliers, while MAE doesn't.

 L2 loss incurs a much higher penalty for an outlier than L1 loss.

 For example, the following images show a **model trained using MAE** and a **model trained using MSE**.

 The red line represents a fully trained model that will be used to make predictions. The outliers are closer to the model trained with MSE than to the model trained with MAE.


 image.png
 MSE loss moves the model closer to the outliers.


 image.png
 MAE loss keeps the model farther from the outliers.



 More guidelines on choosing a loss

 Choose MSE:

If you want to heavily penalize large errors.

If you believe the outliers are important and indicative of true data variance that the model should account for.


    Note: The mathematical properties of MSE often make optimization smoother. Root Mean Squared Error (RMSE) is often used to get the error back into the same units as the label.


Choose MAE:

If your dataset has significant outliers that you don't want to overly influence the model. MAE is more robust.

If you prefer a loss function that is more directly interpretable as the average error magnitude.

In practice, your metric choice can also depend on the specific business problem and what kind of errors are more costly.

# **Gradient Descent**

This is the mathematical technique that alternatively finds the weights and bias that produce the model with the lowest loss.

It finds the best weight and bias by repeating the following processes for a number of iterations.

The model begins training with randomised weights & bias near zero then repeats the following steps

    1. Calculate the loss with the current wight and bias.

    2. Determine the direction to move weights & bias that reduce loss.

    3. Move the weights and bias values a small amount in the direction that reduces loss.

    4. Return to step one and repeat the process until the model cant reduce loss any further.

### **1. Sigmoid Function**

You might be wondering how a logistic regresion model ensures its output represents a probability, always outputing a value between 0 and 1.

We use a family of functions called logistic functions whose outputs has the same characteristics.

The Standard logistic function(sigmoid function) has the following fomula:

$$f(x) = \frac{1}{1 + e^{-x}}$$
where:

* f(x) is the output of the sigmoid function
* e is the euler's number, a mathematical constant 2.71828
* x is the input to the sigmoid function

## **Transforming linear output using the sigmoid function**

The following equation represents the linear component of a logistic regresion model:

$$z = b + w₁x₁ + w₂x₂ + .... wnxn$$

where:
* z is the output of the linear equation(log odds).
* b is the bias.
* The w values are the models learned weights.
* The x values are the feature values for a particullar example.

To obtain the logistic regression prediction, the z value is then passed to the sigmoid function, yielding a probability between o & 1:

$$y' = \frac{1} {1 + e^{-z}}$$

where:
* y' is the output of the logistic regression model
* e is the euler's number a constant 2.71828
* z is the linear output
